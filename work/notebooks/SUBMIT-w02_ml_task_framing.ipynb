## 5. Why ML beats a fixed rule here

**Why ML is better:**
A fixed rule uses rigid buckets (e.g., "If average position is between 4 and 10, the expected CTR is 3%"). But search behavior isn't flat. The CTR drops exponentially from position 4 to position 10. A fixed if-statement treats the top and bottom of Page 1 exactly the same. Machine Learning (like a tree-based regressor) smoothly captures the continuous, non-linear decay between ranking position, impression volume, and expected clicks without needing hundreds of hand-written if-statements.
